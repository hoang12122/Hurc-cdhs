## 2. Phân tách và Tính điểm (Tokenization & TF-IDF Scoring)

```
query = "Mã lỗi E-404 của máy"
print(f"Câu hỏi: {query}")
print(f"Tokens: {tokenize(query)}
")

for doc in MANUAL_PAGES:
    doc_tokens = tokenize(doc['text'])
    score = sum(1 for q in set(tokenize(query)) if q in doc_tokens)
    print(f"Document: {doc['category']}")
    print(f"Match Score: {score}")
    print("-" * 30)
```
